###Changing columns (with some data cleaning)

I want to create 2 sets of columns stating minimum price and maximum price using the information of the Price Range. But here I find a problem that needs data cleaning because the Price Range columns have inconsistent string formats: $, commas, different separators (to; -), ocassional trailing +.

```python
#1. Clean unwanted characters
df['Price Range'] = (
    df['Price Range']
    .astype(str)  # Convert to string to handle non-string or NaN values
    .str.replace(r'[\$,]', '', regex=True)  # remove $ and ,
    .str.replace(r'\+', '', regex=True) # remove +
    .str.strip() # remove trailing spaces
    .str.lower()  # Convert to lowercase for consistent 'to' replacement
)

#2. Split into min and max (handling 'to' and '-')
df[['Minimum Price ($)', 'Maximum Price ($)']] = (
    df['Price Range']
    .str.replace(r'\s*to\s*', '-', regex=True)  # Standardize 'to' (case-insensitive, with spaces) / regex=False because I am replacing plain text to text, not like before the pattern [\$,] that needs regex to be True
    .str.replace(r'\s*-\s*', '-', regex=True)  # Standardize '-' (remove extra spaces)
    .str.split ('-', expand=True) #Split into two columns
)

# 3. Clean up split columns and convert to numeric
# Strip spaces from split columns
df['Minimum Price ($)'] = df['Minimum Price ($)'].str.strip()
df['Maximum Price ($)'] = df['Maximum Price ($)'].str.strip()

# Convert to numeric, coercing errors to NaN
df['Minimum Price ($)'] = pd.to_numeric(df['Minimum Price ($)'], errors='coerce')
df['Maximum Price ($)'] = pd.to_numeric(df['Maximum Price ($)'], errors='coerce')

# Handle invalid or missing prices
# Example: Replace NaN with 0 or drop rows with NaN
df['Minimum Price ($)'] = df['Minimum Price ($)'].fillna(0).astype(int)
df['Maximum Price ($)'] = df['Maximum Price ($)'].fillna(0).astype(int)

# Validate prices (e.g., ensure Minimum <= Maximum and non-negative)
df = df[df['Minimum Price ($)'] >= 0]  # Remove negative minimum prices
df = df[df['Maximum Price ($)'] >= 0]  # Remove negative maximum prices
df = df[df['Minimum Price ($)'] <= df['Maximum Price ($)']]  # Ensure min <= max

#4. Drop intermediate cleaned column (to erase the column 'Price Range' as it would not longer be needed and return a new dataframe)
df.drop('Price Range', axis=1, inplace=True)

df.head()
```

###Create a categorical variable from numerical data (inserting a new column)

```python
df['No. of Surgeons'].dtypes
```

The column 'No. of Surgeons' has integer type as data values. I would like to create a column at the right of it that says a categorical variable. If the number is < 20, the data should read 'Low'; 20-150, 'Intermediate'; >150, 'High'

```python
# Create a categorical column right next to 'No. of Surgeons'
bins = [-float('inf'), 19, 150, float('inf')] # Define ranges: <20, 20-150, >150
labels = ['Low', 'Intermediate', 'High']
df['Surgeon Volume'] = pd.cut(
    df['No. of Surgeons'],
    bins=bins,
    labels=labels,
    right=False,
    include_lowest=True
)

#Reorder again the columns
desired_columns = [
    'City',
    'Procedure',
    'No. of Surgeons',
    'Surgeon Volume',
    'Average Price ($)',
    'Minimum Price ($)',
    'Maximum Price ($)',
    'Top Rated Surgeon',
    'More Information'
]

df = df[desired_columns]

df.head()
```

```python
# Count the number of occurrences for each category in 'Surgeon Volume'
surgeon_volume_counts = df['Surgeon Volume'].value_counts()

# Define the desired order of categories
category_order = ['Low', 'Intermediate', 'High']

# Reindex the Series to the desired order
surgeon_volume_counts = surgeon_volume_counts.reindex(category_order)

# Display the counts
print("Counts of each Surgeon Volume category:")
display(surgeon_volume_counts)
```

####Methods to deal with missing values

There exists methods to deal with missing data from simple to complex that come with pros and cons. Deletions methods and Imputation methods.

#####Deletions Methods

Deletion of rows (Listwise deletion) and deletion of columns (Dropping variables). They are the easiest methods but can lead to a significant loss of valuable information and statistical power, especially if missing values are common.

To simply remove all the rows containig any missing data, use `.dropna()`, and to make the changes permanent to the dataframe use `inplace=True` argument.

Another argument is `axis=`. By default this is set to `0` or `index`, which indicates that we are interested in dropping rows.  When set to `1` or `columns`, columns will be dropped. We can focus which columns to include when checking for missing values `dataframe.dropna(subset=['Name of Column'])`

As stated before, these methods are not very reliable, so I am not going to use it.

#####Imputations methods

**Simple imputations:** Replaces missing values with the mean, median, or mode of that column.



*   **Mean Imputation:** Best for numerical values with a normal distribution and no significant otliers.
*   **Median Imputation:** For numerical data that is skewed or contains significant outliers.

*   **Mode Imputation:** A good approach for categorical variables

**Forward and backward fill:** Replaces missing values using the value from the previous observation (forward fill) or the next observation (backward fill). Ideal for time-series data, where you can assume that the value will remain constant over a short period. To fill the Nan data with another value use `.fillna()`, and then the argument `method=ffill` (forward fill) uses the previous valid value to replace the Nan, and `method=bfill` (backward fill) uses the next valid value.

*   dataframe['Column'].fillna(method='ffill')



**K-Nearest Neighbors (KNN) imputation:** Uses a specified number of "neighboring" observations to impute missing values based on similarity. For a missing value, it finds the 'k' closest data points in the dataset and uses a weighted average of their values to make a guess.
This is for complex analysis, and can handle both numerical and categorical data


**Advanced imputation:**

*   **Regression imputation:** Uses a regression model to predict missing values based on other variables in the dataset.
*   **Multiple Imputation by Chained Equations (MICE):** A sophisticated method that creates multiple imputed datasets.

Once a very rapid review of imputation, I will use simple imputation with median

```python
#Median calculation for Minimum Price ($)
median_min_price = df.loc[df['Minimum Price ($)'] > 0, 'Minimum Price ($)'].median() # I used > 0 to exclude 0 from the median formula
median_min_price
```

```python
#Median calculation for Maximum Price ($)
median_max_price = df.loc[df['Maximum Price ($)'] > 0, 'Maximum Price ($)'].median() # I used > 0 to exclude 0 from the median formula
median_max_price
```

```python
#Replace zeros with the median (only when Avg > 0) for Minimum Price ($)
df.loc[(df['Minimum Price ($)'] == 0) & (df['Average Price ($)'] > 0), 'Minimum Price ($)'] = median_min_price
```

```python
#Replace zeros with the median (only when Avg > 0) for Maximum Price ($)
df.loc[(df['Maximum Price ($)'] == 0) & (df['Average Price ($)'] > 0), 'Maximum Price ($)'] = median_max_price
```
